Data Integrity Report:

 1-Missing Values Problem:

  What I found: I found alot missing values in this data

  Where It was: It was in 15 columns which are first_name, last_name, grade, neighborhood, membership_status, join_date, checkout_id, book_id, checkout_date, return_date, borrowed_books, genre, pages, publication_year, publisher   

  How big It was: The affected rows varies by column (first_name:26 rows, last_name:26 rows, grade:64 rows, neighborhood:26 rows, membership_status:26 rows, join_date: 32 rows, checkout_id:44 rows, book_id:18 rows, checkout_date:18 rows, return_date:109 rows, borrowed_books:26 rows, genre:44 rows, pages:44 rows, publication_year:78 rows, publisher:44 rows)

  What I did and why: 
   I fill the missing values in first_name, last_name, checkout_id, book_id and publisher columns with "unknown", because dropping these rows would lead to large data loss
   The missing values in return_date column were filled as "Not Returned" because the books are still checked out by members

   I fill the missing values in grades, neighborhood, membership_status, join_date, checkout_date, genre and publication_year with the mode because it represent the most frequent value

   I fill the missing values in the pages column with the median because the outliers don't affect it and as it contain numeric data

   I fill the missing values in borrowed_books column with 0 because the members didn't borrow any books untill now

2-Duplicates Problem:

 What I found: I found true duplicates that are all the same

 Where It was: Across all columns as entire rows were duplicated

 How big It was: The affected rows are 8

 What I did and why: I dropped all duplicated rows to ensure that every row in the data is unique 

3-The same values written in many ways problem (Inconsistent Data):

 What I found: I made a check on every column contain categorical data and I found inconsistent data

 Where It was: It was in 2 column which are neighborhood and membership_status columns

 How big It was: The affected rows are 31 rows in the neighborhood column ('Maadi', 'Maadi ', 'Nasr City', 'Nasr  City', 'NASR CITY','HELIOPOLIS', 'Heliopolis', 'zamalek', 'Zamalek', 'Shubra') and 77 rows in the membership_status column ('Active', 'inactive', 'Inactive', 'active', 'INACTIVE)

 What I did and why: For both, I removed the leading and trailing space and converting them to title case, in the neighborhood column, I replaced multiple spaces with a single space to make pandas group them correctly without creating duplicate categories

4-Checkouts with no matching member problem:

 What I found : I found that some rows contain member_id that doesn't actually exist among the registered members

 Where It was: It was in member_id column

 How big It was: The affected rows are 5 rows

 What I did and why: I dropped these 5 rows because their member_id is not registered in the members table and they represent invalid checkouts
